# Anki - Kubernetes labels, tains, tolerations

```python
import pandas as pd
import markdown
import urllib.request
import re
```

```python
url = "https://raw.githubusercontent.com/kubernetes/website/main/content/en/docs/reference/labels-annotations-taints/_index.md"
with urllib.request.urlopen(url) as response:
    md = response.read().decode("utf-8")
```

```python
md
```

```python
no_meta = re.sub("^---.*---\n\n", "", md, flags=re.DOTALL)
```

```python
html = markdown.markdown(no_meta, extensions=["fenced_code"])
```

```python
forward = {}
backward = {}
for g in html.split("<h3>"):
    g = g.split("</h3>")
    if len(g) < 2:
        print(g)
        continue
    title = g[0]
    body = " ".join(g[1].split("\n")).strip()
    summary = body.replace("Example", "Used on").split("Used on")
    if len(summary) < 2:
        summary = body
    else:
        summary = summary[-1].split("</p>")[1].strip()

    forward[title] = body
    backward[summary] = f"{title}\n{body}"
```

```python
forward_df = pd.DataFrame(data=forward.values(), index=forward.keys())
backward_df = pd.DataFrame(data=backward.values(), index=backward.keys())
df = pd.concat([forward_df, backward_df])
df
```

```python
df.to_csv(
    "../../.ignore/kube-labels.csv", header=False, encoding="utf-8", sep=";"
)
```
