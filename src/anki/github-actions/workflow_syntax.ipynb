# Github actions workflow syntax

```python
from os import environ, path, makedirs
from bs4 import BeautifulSoup, NavigableString, PageElement
from IPython.display import display
from IPython.core.display import HTML
import pandas as pd
from markupsafe import escape

assets_abspath = path.join(environ.get("ASSETS_ABSPATH", ""), "github-actions")
artifacts_abspath = path.join(
    environ.get("ARTIFACTS_ABSPATH", ""), "github-actions"
)
source_html_abspath = path.join(assets_abspath, "workflow-syntax.html")
target_csv_abspath = path.join(artifacts_abspath, "workflow-syntax.csv")
```

```python
makedirs(artifacts_abspath, exist_ok=True)
```

```python
with open(source_html_abspath, "r") as f:
    html = f.read()
html
```

```python
DOMAIN = "https://docs.github.com"
ROUTE = "/en/actions/using-workflows/workflow-syntax-for-github-actions"

soup = BeautifulSoup(html, "html.parser")
for anchor in soup.find_all("a"):
    if anchor["href"].startswith("#"):
        anchor["href"] = DOMAIN + ROUTE + anchor["href"]
    elif anchor["href"].startswith("/"):
        anchor["href"] = DOMAIN + anchor["href"]

sections = []
current = None
for child in soup.children:
    if isinstance(child, NavigableString):
        continue
    # if isinstance(child, PageElement):
    #     print("page element!")
    #     continue
    if child.name == "h2":
        if current is not None:
            sections.append(current)
        current = {
            "header_soup": child,
            "content_soup": BeautifulSoup("", "html.parser"),
        }
        continue
    if current is not None:
        current["content_soup"].append(child)
else:
    sections.append(current)


enriched = []
for section in sections:
    header_text = section["header_soup"].text
    header_text = header_text.replace("<", "&lt;").replace(">", "&gt;")
    content_text = section["content_soup"].find("p").text
    content_text = content_text.replace("<", "&lt;").replace(">", "&gt;")

    summary_text_arr = content_text.split(". ")
    summary_text = summary_text_arr[0]
    if len(summary_text) < 30:
        summary_text = ". ".join(summary_text_arr[0:2])
    if not summary_text.endswith("."):
        summary_text += "."
    summary_text = summary_text.replace(header_text, "___")

    description_soup = BeautifulSoup(
        section["header_soup"].prettify() + section["content_soup"].prettify(),
        "html.parser",
    )

    enriched.append(
        {
            # "header_h2_html": section["header_soup"].prettify(),
            "header_html": f"<p>{header_text}</p>",
            "description_html": description_soup.prettify(),
            "summary_html": f"<p>{summary_text}</p>",
            "tags": "WebScraped GithubActions-2024-07-16",
        }
    )

for item in enriched:
    display(HTML(item["header_html"]))
    # display(HTML(item["summary_html"]))
    display(HTML(item["description_html"]))
    print("-" * 50)
```

```python
df = pd.DataFrame(enriched)
df.rename(
    columns={
        "header_html": "HeaderHtml",
        "description_html": "DescriptionHtml",
        "summary_html": "SummaryHtml",
    },
    inplace=True,
)
df
```

```python
df.to_csv(
    target_csv_abspath, sep="|", index=False, header=False, encoding="utf-8"
)
```
