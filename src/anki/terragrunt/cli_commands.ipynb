```python
from bs4 import BeautifulSoup
from src.anki.terragrunt.common import (
    get_page,
    write_html,
    read_soup,
    fix_hrefs,
    parse_entries,
    enrich_entries,
    create_df,
)
from os.path import join
import re
import pandas as pd

BASE = "https://terragrunt.gruntwork.io"
REPO_ABSPATH = "/utkusarioglu-com/workshops/backtesting-workshop"
ARTIFACTS_SUBPATH = "artifacts/terragrunt"

artifacts_abspath = join(REPO_ABSPATH, ARTIFACTS_SUBPATH)
artifact_file_abspath = join(artifacts_abspath, "terragrunt.html")
```

```python
soup = read_soup(artifact_file_abspath)
```

```python
article = soup.find(class_="cb-doc-content js-scroll-spy")
article
```

```python
fix_hrefs(BASE, article)
```

```python
entries = parse_entries(article)
```

```python
content = enrich_entries(entries)
```

```python
cli_commands = create_df(content[0]["list"])
cli_commands.to_csv(
    join(artifacts_abspath, "cli_commands.csv"),
    encoding="UTF8",
    header=False,
    index=False,
    sep=":",
)
```

```python
cli_options = create_df(content[1]["list"])
cli_options.to_csv(
    join(artifacts_abspath, "cli_options.csv"),
    encoding="UTF8",
    header=False,
    index=False,
    sep=":",
)
```
