# Terragrunt Built-in Functions

```python
from os.path import join
from requests import get
from bs4 import BeautifulSoup
from copy import deepcopy
import re
import pandas as pd

BASE = "https://terragrunt.gruntwork.io"
REPO_ABSPATH = "/utkusarioglu-com/workshops/backtesting-workshop"
ARTIFACTS_SUBPATH = "artifacts/terragrunt"
url = f"{BASE}/docs/reference/built-in-functions"
artifacts_abspath = join(REPO_ABSPATH, ARTIFACTS_SUBPATH)
artifact_html_abspath = join(artifacts_abspath, "built_in_functions.html")
artifact_csv_abspath = join(artifacts_abspath, "built_in_functions.csv")
```

```python
# Fetch and store html artifact
def fetch_and_store():
    response = get(url)
    if response.status_code != 200:
        raise RuntimeError("FETCH_UNSUCCESSFUL")
    html = open(artifact_html_abspath, "w+")
    html.write(response.text)
    html.close()


# fetch_and_store()
```

```python
cache = open(artifact_html_abspath, "r")
html = cache.read()
cache.close()
soup = BeautifulSoup(html)
soup
```

### Find article

```python
article = soup.find(class_="cb-doc-content js-scroll-spy")
article
```

### Fix href and src attributes

```python
hrefs = article.find_all(lambda e: e.has_attr("href"))
for href in hrefs:
    if href["href"].startswith("/"):
        href["href"] = BASE + href["href"]
    elif href["href"].startswith("#"):
        href["href"] = url + "/" + href["href"]
srcs = article.find_all(lambda e: e.has_attr("src"))
for src in srcs:
    if src["src"].startswith("/"):
        src["src"] = BASE + src["src"]
article
```

### Divide to sections

```python
h2_curr = None
sections = []
children = BeautifulSoup("", "html.parser")
for elem in deepcopy(article).find_all(recursive=False):
    if elem.name == "h2":
        if h2_curr is not None:
            sections.append({"h2": h2_curr, "children": children})
        h2_curr = elem
        children = BeautifulSoup("", "html.parser")
        continue
    children.append(elem)

sections
```

```python
enriched = []
for section in sections:
    title = section["h2"].string
    summary = section["children"].find("p").text
    summary = re.split(r"\.\s", summary)[0] + "."
    summary = summary.replace(title, "___")
    # details_with_header = BeautifulSoup(title.prettify() + section["children"].prettify(), "html.parser")
    details = section["children"].prettify()
    details_with_header = BeautifulSoup(
        section["h2"].prettify() + details,
        "html.parser",
    ).prettify()

    enriched.append(
        {
            **section,
            "title": title,
            "summary": summary,
            "details": details,
            "details_with_header": details_with_header,
            # "tags": [],
        }
    )

enriched
```

```python
functions = pd.DataFrame(enriched)[
    ["title", "summary", "details_with_header", "details"]
]
functions
```

```python
forwards = functions[["title", "details"]].copy()
forwards = forwards.rename(columns={"title": "front", "details": "back"})
backwards = functions[["summary", "details_with_header"]].copy()
backwards = backwards.rename(
    columns={"summary": "front", "details_with_header": "back"}
)
combined = pd.concat([forwards, backwards], ignore_index=True)
combined
```

```python
combined.to_csv(artifact_csv_abspath, sep=":", index=False, header=False)
```
