```python
from requests import get
from bs4 import BeautifulSoup
import pandas as pd
from os.path import join
from markdown_it import MarkdownIt

URL = "https://raw.githubusercontent.com/kubernetes/website/main/content/en/docs/reference/labels-annotations-taints/_index.md"
ARTIFACTS_RELPATH = "../../../artifacts"
cache_relpath = join(ARTIFACTS_RELPATH, "kubernetes/labels.md")
```

```python
def cache_source():
    response = get(URL)
    if response.status_code != 200:
        raise RuntimeError("FETCH_FAILED")
    with open(cache_relpath, "w") as f:
        f.write(response.text)


# cache_source()
```

```python
def fetch_cache():
    with open(cache_relpath, "r") as f:
        content = f.read()
    md = MarkdownIt()
    html = md.render(content)
    return BeautifulSoup(html, "html.parser")


article = fetch_cache()
article
```
